```python
import pandas as pd, numpy as np
matrix = np.random.randint(0, 5, 25).reshape(5,5).astype(float)
matrix[2,2] = None
df = pd.DataFrame(matrix, columns=['A', 'B', 'C', 'D', 'E'])
print(df)
print(df.max())
```

```python
df.mean()
```

```python
print(df.describe())
print(df.describe().T)
```

```python
df.isna().sum()
```

```python
print(f"Кол-во уникальных значений в колонке B: {df["B"].nunique()}")
print(df["B"].unique())
```

```python
print("Удаление строк с na")
print(df.dropna())
```

```python
print("Удаление столбцов с na")
print(df.dropna(axis=1))
```

```python
print("Заменяю значние nan на среднее значения колонки")
print(df.fillna(df.mean()))
```

```python
d = {'Company':['COMP1', 'COMP1', 'COMP2', 'COMP2', 'COMP3', 'COMP3'],
         'Month':['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN'],
         'Salary':[2000, 1500, 3000, 5000, 2500, 3500]}
df2 = pd.DataFrame(d)
print(df2)
```

```python
df2.groupby('Company').agg({'Salary': ['sum'], 'Month':['count']})
```

```python
df2.groupby('Company').agg({'Salary':['mean', 'max', 'min']})
```

```python
df2.groupby('Company').agg({'Salary':['count']})
```

```python
df2.groupby('Company').describe().transpose()['COMP3']
```
