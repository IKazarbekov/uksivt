### Задания по NumPy

1. Создайте массив из списка, состоящего из чисел от 0 до 9, при помощи функции array() из модуля numpy.

2. Создайте массив, состоящий из нечётных чисел от 1 до 19, при помощи функции arange() из модуля numpy.

3. Создайте массив, состоящий из двадцати нулей.

4. Создайте двумерный массив, состоящий из двадцати единиц, в котором 5 строк и 4 столбца.

5. Создайте массив, состоящий из пяти равномерно распределённых чисел от 3 до 7.

6. Создайте двумерный массив, состоящий из 9 равномерно распределённых чисел от 3 до 7.

7. Создайте массив, состоящий из 10 случайных целых чисел из диапазона от 1 до 100. Присвойте его максимальное значение переменной max_number и выведите значение этой переменной.

8. Создайте массив, состоящий из 10 случайных целых чисел из диапазона от 1 до 20. Сформируйте новый массив, состоящий из 5 элементов первого массива, начиная со 2 по порядку. Выведите значения обоих массивов.

9. Создайте двумерный массив, состоящий из случайных целых чисел из диапазона от 3 до 11, в котором 4 строки и 3 столбца. Сформируйте новый массив, состоящий из элементов  второй и третьей строки, первого и второго столбца по порядку. Выведите значения обоих массивов при помощи функции print().

10. Создайте массив, состоящий из 10 случайных целых чисел из диапазона от 0 до 20, затем создайте массив состоящий из элементов, начиная с третьего по порядку до последнего. Выведите значения обоих массивов при помощи функции print().

11. Создайте двумерный массив, состоящий из случайных целых чисел из диапазона от 0 до 9, в котором 4 строки и 6 столбцов.  Создайте новый массив, состоящий из элементов, больших числа 3. Выведите значения обоих массивов при помощи функции print().

12. Создайте массив, состоящий из 10 случайных целых чисел из диапазона от 1 до 20. Cоздайте новый массив, состоящий из 10 элементов, начиная с 2. Выведите значения этих массивов, а также их сумму.

13. Создайте массив, состоящий из 7 случайных целых чисел из диапазона от 0 до 9. Создайте новый массив, состоящий из 7 элементов первого массива, увеличенных в 3 раза. Выведите значения этих массивов при помощи функции print().

14. Создайте массив, состоящий из квадратных корней 20 элементов по порядку, начиная с 7. Выведите значение массива при помощи функции print().

15. Даны x = [1,2], y = [[4, 1], [2, 2]]. Найдите произведение матриц x * y и y * x.

16. Даны x = [[1, 0], [0, 1]], y = [[4, 1], [2, 2], [1, 1]]. Найдите произведение матриц.

17. Создайте массив x, состоящий из 10 чисел от 0 до 9. Если элемент x меньше 3, замените его на 3. И если элемент x больше 7, замените его на 7. 

18. Создайте массив x = np.array([1, 3, 0, -1, -3]). Вычислите поэлементно знак x (sign x).

19. Дана матрица M = [[0 5 9 8 2], [3 7 4 1 6]]. Найдите максимальное и минимальное значения и их индексы по строкам и столбцам.

20. Создайте массив размерности 3 на 3, состоящий из чисел от 1 до 9. Вычислите среднее значение, медиану, стандартное отклонение и дисперсию.
